```python
from langchain_community.tools import DuckDuckGoSearchRun,WikipediaQueryRun
from langchain_experimental.tools import PythonREPLTool
from langchain_community.utilities import WikipediaAPIWrapper
from langchain.agents import create_agent
from langchain_groq import ChatGroq 
from dotenv import load_dotenv
load_dotenv()
```

# DuckDuckGoSearchRun: 
### Web serach via DuckDuckGoSearch

```python
search_tool=DuckDuckGoSearchRun()
result=search_tool.invoke("Dar-e-Nijaat")
print(result)
```

```python
print(search_tool.args) #to see the parameters or arguments and their type
print(search_tool.description) # to see the description
```

# WikipediaQueryRun:
### API WRAPPPER = it is used so that llm can communicate to wikipedia or any other external source and fetch data from them.. so we have to use WikipediaAPIWrapper in this case

```python
api=WikipediaAPIWrapper(
    top_k_results=3,
    doc_content_chars_max=2000
)
```

```python
wiki_tool = WikipediaQueryRun(api_wrapper=api)
result = wiki_tool.invoke("Sheheryar Munawar")
print(result)
```

# PythonREPLTools
### Writes raw python code only and only when llm is present ..

```python
llm= ChatGroq(model="openai/gpt-oss-20b")

python_tool = PythonREPLTool()

agent=create_agent(
    model=llm,
    tools=[python_tool]
)

result=agent.invoke({"messages":[
    {
        "role":"user",
        "content":"write simple code for checking whether the given number is prime or not"
    }
]})

print(result["messages"][-1].content)
```
