# WikipediaQueryRun:
### API WRAPPPER = it is used so that llm can communicate to wikipedia or any other external source and fetch data from them.. so we have to use WikipediaAPIWrapper in this case

```python
api=WikipediaAPIWrapper(
    top_k_results=3,
    doc_content_chars_max=2000
)
```

```python
wiki_tool = WikipediaQueryRun(api_wrapper=api)
result = wiki_tool.invoke("Sheheryar Munawar")
print(result)
```
