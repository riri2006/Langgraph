```python
from langchain_community.tools import tool
```

# Custom Tools: 

### @TOOL makes the function a runnable tool so that it can run by using invoke() 

```python
@tool
def Multiply(a:int,b:int) ->int:
    """Multiply the two integers and give their product"""
    return a*b
```

```python
result= Multiply.invoke({"a":12,"b":15})
print("Product:",result)
```

```python
# To check the given name , description and arguments of tool:::------
print("NAME:",Multiply.name)
print("DESCRIPTION:",Multiply.description)
print("ARGUMENTS:",Multiply.args)
```
